```python
df = df[~(df['clean_comment'].str.strip() == '')]
```

```python
df['clean_comment'] = df['clean_comment'].str.strip()
df['clean_comment'].apply(lambda x: x.endswith(' ') or x.startswith(' ')).sum()
```

```python
# Removing URLs from comments
url_pattern = r"^(https?):\/\/([a-zA-Z0-9.-]+)(:[0-9]+)?(\/[^\s]*)?$"

comments_with_urls = df[df['clean_comment'].str.contains(url_pattern, regex=True)]

comments_with_urls.head()
```

```python
# is there any new lines in comments
newlines_in_comments = df[df['clean_comment'].str.contains('\n')]
newlines_in_comments.head()
```

```python
# Removing newline (\n) from comments
df['clean_comment'] = df['clean_comment'].str.replace('\n', ' ', regex=True)
```

```python
# is there any new lines in comments
newlines_in_comments = df[df['clean_comment'].str.contains('\n')]
newlines_in_comments.head()
```
